# VenuraTech Programs Section Analysis

This notebook fetches, parses, and displays program data from [venuratech.com/programs](https://venuratech.com/programs), and provides a design reference for implementing a similar section in the current Venura codebase.

## 1. Import Required Libraries

We will use `requests` for HTTP requests, `BeautifulSoup` for HTML parsing, `pandas` for data manipulation, and `matplotlib` for visualization.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Fetch Program Data from venuratech.com

We will retrieve the HTML content from the programs page.

```python
url = 'https://venuratech.com/programs'
response = requests.get(url)
html_content = response.text
print(f'Fetched {len(html_content)} characters of HTML.')
```

## 3. Parse and Extract Program Information

We will use BeautifulSoup to parse the HTML and extract program details such as title, description, category, and duration.

```python
soup = BeautifulSoup(html_content, 'html.parser')

# Example: Find all program cards (update selector as needed)
programs = []
for card in soup.select('.program-card, .program, .course-card'):
    title = card.select_one('.title, .program-title, h3')
    desc = card.select_one('.desc, .program-desc, p')
    category = card.select_one('.category, .program-category')
    duration = card.select_one('.duration, .program-duration')
    programs.append({
        'Title': title.text.strip() if title else '',
        'Description': desc.text.strip() if desc else '',
        'Category': category.text.strip() if category else '',
        'Duration': duration.text.strip() if duration else ''
    })

print(f'Extracted {len(programs)} programs.')
```

## 4. Display Program Data in Tabular Format

We will use pandas to organize and display the extracted program data in a table.

```python
df = pd.DataFrame(programs)
display(df.head())
```

## 5. Visualize Program Categories

We will use matplotlib to create a bar chart showing the number of programs in each category.

```python
if not df.empty and 'Category' in df.columns:
    category_counts = df['Category'].value_counts()
    plt.figure(figsize=(8, 5))
    category_counts.plot(kind='bar', color='#FF7A00')
    plt.title('Number of Programs by Category')
    plt.xlabel('Category')
    plt.ylabel('Number of Programs')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
else:
    print('No category data available to plot.')
```
